The Gaussian Elimination method transforms the system of equations into an upper triangular matrix and then uses backward substitution to solve for the variables.

```python
import numpy as np

def gaussian_elimination(A, b):
	"""
	Solves the system Ax = b using Gaussian Elimination with partial pivoting.
    
	:param A: Coefficient matrix
	:param b: Right-hand side vector
	:return: Solution vector x
	"""
	for pidx in range(len(A)-1):
		highest_index = (None, None)
		current_x = current_y = pidx
		for index, values in enumerate(A[current_x:]):
			if highest_index[0] == None or values[current_y] > highest_index[1]:
				highest_index = (index + current_x, values[current_y])
		A[[current_x, highest_index[0]]] = A[[highest_index[0], current_x]]
		b[[current_x, highest_index[0]]] = b[[highest_index[0], current_x]]
		pivot_value = A[current_x][current_y]
		values_to_pivot_upon = A[current_x]
		for index, values in enumerate(A[current_x + 1:]):
			m = values[current_x] / pivot_value
			for i in range(len(values_to_pivot_upon)):
				values[i] -= m * values_to_pivot_upon[i]
			b[index + 1 + current_x] = b[index + 1 + current_x] - m * b[current_x]
			A[index + 1 + current_x] = values

		print(A)
		print(b)


	results = np.zeros_like(b)
	for pidx in range(len(A)):
		current_idx = len(A) - pidx - 1
		a_row = A[current_idx]
		b_val = b[current_idx]
		numerator = b_val - sum(results[current_idx:] * a_row[current_idx:])
		results[current_idx] = numerator / a_row[current_idx]

	return results

A = np.array([[2,8,4], [2,5,1], [4,10,-1]], dtype=float)
b = np.array([2,5,1], dtype=float)

print(gaussian_elimination(A, b)) # [11.0, -4.0, 3.0]
```
