```python
import numpy as np

def get_all_but_indx(r, idx):
	mask=np.ones_like(r, bool)
	mask[idx]=False
	return r[mask]

def get_subtraction(vals, a_row, idx):
	oth_val = get_all_but_indx(vals, idx)
	oth_a_row = get_all_but_indx(a_row, idx)
	return sum(np.multiply(oth_val, oth_a_row))

def gauss_seidel(A, b, n, x_ini=None):
	values = np.zeros_like(b)
	if x_ini:
		values[:] = x_ini
	while n > 0:
		for idx, a_v in enumerate(A):
			values[idx] = (b[idx] - get_subtraction(values, a_v, idx)) / a_v[idx]
			if idx == len(A) - 1:
				n -= 1
	return values
```

```python
A = np.array([[4, 1, 2], [3, 5, 1], [1, 1, 3]], dtype=float)
b = np.array([4, 7, 3], dtype=float)

n = 100
print(gauss_seidel(A, b, n))
```
